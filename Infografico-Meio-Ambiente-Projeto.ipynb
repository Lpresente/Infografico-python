```python
"""Infográfico ilustrado: Meio ambiente — desenvolvimento sustentável.
Execute: python infografico_sustentabilidade_v2.py
As fontes estão incorporadas e todas as imagens são desenhadas em Python.
"""
from pathlib import Path
from PIL import Image, ImageDraw, ImageFont
import argparse
import base64
import io
import math
import random
```

```python
# 1. Fontes incorporadas. Isso evita arquivos extras e diferenças entre máquinas.
FONT_TITULO = ""
```

```python
FONT_TEXTO = ""
```

```python
# 2. Cores, tamanho e prancheta. Cada cor tem uma função no desenho.
C = dict(papel="#FFFFFF", tinta="#173E37", verde="#28865F", floresta="#145B44",
         menta="#EAF5EC", lima="#B4D46B", azul="#2789AC", ceu="#E8F4F7",
         rio="#6DBBD1", coral="#DA7659", rosa="#FFF0E9", ouro="#F2C75A",
         suave="#526C65", linha="#D8E6DF", claro="#F6F9F7")
W, H = 1600, 2480
im = Image.new("RGB", (W, H), C["papel"])
d = ImageDraw.Draw(im)
random.seed(18)  # Repetibilidade do desenho; não representa uma data.
FONTES = {}
```

```python
def fonte(tamanho, destaque=False):
    chave=(tamanho,destaque)
    if chave not in FONTES:
        dados=base64.b64decode(FONT_TITULO if destaque else FONT_TEXTO)
        FONTES[chave]=ImageFont.truetype(io.BytesIO(dados), tamanho)
    return FONTES[chave]
```

```python
def txt(x,y,s,tam=26,cor="tinta",bold=False,anchor="lt"):
    """Texto com posição e tamanho fixos; aceita quebras de linha."""
    for i, trecho in enumerate(s.split("\n")):
        d.text((x,y+i*(tam+10)),trecho,font=fonte(tam,bold),
               fill=C.get(cor,cor),anchor=anchor)
```

```python
def rect(box, cor): d.rectangle(box, fill=C.get(cor,cor))
def poly(p,cor): d.polygon(p,fill=C.get(cor,cor))
def line(p,cor="tinta",width=3): d.line(p, fill=C.get(cor,cor),width=width)
def circle(x,y,r,cor): d.ellipse((x-r,y-r,x+r,y+r),fill=C.get(cor,cor))
def ellipse(box,cor): d.ellipse(box,fill=C.get(cor,cor))
def seta(a,b,cor="suave",width=3):
    line([a,b],cor,width)
    ang=math.atan2(b[1]-a[1],b[0]-a[0])
    poly([b,(b[0]-13*math.cos(ang-.45),b[1]-13*math.sin(ang-.45)),
          (b[0]-13*math.cos(ang+.45),b[1]-13*math.sin(ang+.45))],cor)
```

```python
def leaf(x,y,s=1,cor="verde"):
    pts=[(x,y+25*s),(x-26*s,y+7*s),(x-20*s,y-19*s),
         (x+16*s,y-35*s),(x+27*s,y-9*s),(x+12*s,y+15*s)]
    poly(pts,cor)
    line([(x-9*s,y+20*s),(x+15*s,y-23*s)],"papel",max(2,int(2*s)))
```

```python
def tree(x,y,s=1,cor="verde"):
    rect((x-3*s,y-62*s,x+3*s,y),"floresta")
    ellipse((x-26*s,y-106*s,x+26*s,y-36*s),cor)
    line([(x,y-38*s),(x,y-86*s)],"floresta",max(2,int(s*2)))
    line([(x,y-55*s),(x-14*s,y-69*s)],"floresta",max(2,int(s*2)))
```

```python
def person(x,y,s=1,cor="coral"):
    circle(x,y-40*s,7*s,"tinta")
    line([(x,y-29*s),(x,y-5*s)],cor,max(4,int(s*12)))
    line([(x,y-8*s),(x-8*s,y+13*s)],"tinta",max(3,int(s*4)))
    line([(x,y-8*s),(x+8*s,y+13*s)],"tinta",max(3,int(s*4)))
    line([(x-13*s,y-13*s),(x,y-27*s),(x+13*s,y-12*s)],cor,max(3,int(s*4)))
```

```python
def house(x,y,s=1):
    rect((x,y-55*s,x+65*s,y),"papel")
    poly([(x-6*s,y-55*s),(x+32*s,y-82*s),(x+71*s,y-55*s)],"coral")
    rect((x+26*s,y-31*s,x+40*s,y),"tinta")
    rect((x+8*s,y-43*s,x+20*s,y-30*s),"azul")
    rect((x+48*s,y-43*s,x+58*s,y-30*s),"azul")
```

```python
def building(x,y,w,h):
    rect((x,y-h,x+w,y),"papel")
    rect((x+w-10,y-h,x+w,y),"linha")
    for xx in range(int(x+10),int(x+w-12),22):
        for yy in range(int(y-h+15),int(y-12),25):
            rect((xx,yy,xx+9,yy+12),"azul")
    rect((x-2,y-h-6,x+w+2,y-h),"verde")
    for xx in range(int(x+6),int(x+w),14): circle(xx,y-h-8,7,"lima")
```

```python
def solar(x,y,s=1):
    poly([(x,y),(x+86*s,y),(x+103*s,y+43*s),(x+17*s,y+43*s)],"azul")
    for i in range(1,4):
        line([(x+i*21.5*s,y+3*s),(x+i*21.5*s+15*s,y+40*s)],"ceu",2)
    line([(x+9*s,y+21*s),(x+93*s,y+21*s)],"ceu",2)
    line([(x+28*s,y+43*s),(x+28*s,y+57*s)],"tinta",3)
    line([(x+86*s,y+43*s),(x+86*s,y+57*s)],"tinta",3)
```

```python
def turbine(x,y,s=1):
    line([(x,y),(x,y-123*s)],"papel",max(3,int(s*7)))
    circle(x,y-123*s,6*s,"tinta")
    for a in [-90,30,150]:
        theta=math.radians(a)
        line([(x,y-123*s),(x+56*s*math.cos(theta),y-123*s+56*s*math.sin(theta))],
             "papel",max(3,int(5*s)))
```

```python
def bike(x,y,s=1):
    for dx in [-29,29]:
        d.ellipse((x+(dx-20)*s,y-20*s,x+(dx+20)*s,y+20*s),outline=C["tinta"],width=max(2,int(3*s)))
    line([(x-29*s,y),(x-9*s,y-32*s),(x+10*s,y),(x-29*s,y),
          (x+29*s,y),(x+14*s,y-38*s)],"coral",max(3,int(s*4)))
    line([(x-15*s,y-34*s),(x-4*s,y-34*s)],"tinta",3)
```

```python
def icon(x,y,kind,s=1,cor="verde"):
    """Ícones originais feitos com formas básicas."""
    if kind=="folha": leaf(x,y,s,cor)
    elif kind=="pessoas":
        for dx,dy in [(-26,9),(26,9),(0,-7)]:
            circle(x+dx*s,y+(dy-16)*s,9*s,cor)
            ellipse((x+(dx-13)*s,y+dy*s,x+(dx+13)*s,y+(dy+31)*s),cor)
    elif kind=="economia":
        circle(x,y,32*s,cor)
        txt(x,y-18*s,"$",int(40*s),"papel",True,anchor="mt")
        leaf(x+35*s,y-24*s,.55*s,"verde")
    elif kind=="agua":
        poly([(x,y-37*s),(x-25*s,y+1*s),(x-22*s,y+23*s),
              (x,y+34*s),(x+22*s,y+23*s),(x+25*s,y+1*s)],cor)
        line([(x-13*s,y+6*s),(x-10*s,y+20*s)],"papel",max(2,int(4*s)))
    elif kind=="energia":
        poly([(x+8*s,y-37*s),(x-26*s,y+5*s),(x-3*s,y+5*s),
              (x-10*s,y+37*s),(x+27*s,y-8*s),(x+4*s,y-8*s)],cor)
    elif kind=="bike": bike(x,y,s)
    elif kind=="repensar":
        d.ellipse((x-28*s,y-34*s,x+28*s,y+18*s),outline=C[cor],width=max(3,int(5*s)))
        line([(x-11*s,y+18*s),(x-11*s,y+30*s),(x+11*s,y+30*s),(x+11*s,y+18*s)],cor,max(3,int(5*s)))
        for ang in [-150,-90,-30]:
            a=math.radians(ang)
            line([(x+39*s*math.cos(a),y-7*s+39*s*math.sin(a)),
                  (x+48*s*math.cos(a),y-7*s+48*s*math.sin(a))],cor,3)
    elif kind=="recusar":
        rect((x-22*s,y-20*s,x+22*s,y+30*s),cor)
        d.arc((x-12*s,y-35*s,x+12*s,y-5*s),180,360,fill=C[cor],width=4)
        line([(x-32*s,y+38*s),(x+33*s,y-35*s)],"papel",max(3,int(5*s)))
    elif kind=="reduzir":
        for i,h in enumerate([55,39,23]): rect((x-37*s+i*27*s,y+28*s-h*s,x-19*s+i*27*s,y+28*s),cor)
        seta((x-30*s,y-42*s),(x+34*s,y-20*s),cor,4)
    elif kind=="reutilizar":
        rect((x-22*s,y-16*s,x+22*s,y+28*s),cor)
        rect((x-26*s,y-26*s,x+26*s,y-18*s),cor)
        d.arc((x-43*s,y-44*s,x+43*s,y+44*s),-40,155,fill=C[cor],width=4)
        poly([(x-40*s,y+8*s),(x-48*s,y+25*s),(x-30*s,y+20*s)],cor)
    elif kind=="reciclar":
        pts=[(x,y-33*s),(x+34*s,y+26*s),(x-34*s,y+26*s)]
        for i in range(3):
            a,b=pts[i],pts[(i+1)%3]
            a=(a[0]*.85+x*.15,a[1]*.85+y*.15)
            b=(b[0]*.85+x*.15,b[1]*.85+y*.15)
            seta(a,b,cor,max(3,int(5*s)))
```

```python
# 3. Título editorial e conceito, com texto enxuto.
print("1/4 • Criando título e organizando as informações...")
txt(80,53,"MEIO AMBIENTE  /  FUTURO EM EQUILÍBRIO",24,"verde",True)
txt(75,102,"Desenvolvimento",104,bold=True)
txt(75,212,"sustentável",116,"verde",True)
leaf(1436,185,2.25)
txt(80,350,"Viver bem hoje, sem tirar das próximas gerações\na possibilidade de viver bem amanhã.",31)
```

```python
# 4. Paisagem ilustrada: relações entre natureza, cidade e energia.
print("2/4 • Desenhando a paisagem, a cidade e as energias renováveis...")
rect((0,462,1600,942),"ceu")
circle(1358,530,46,"ouro")
for x,y,s in [(210,510,1),(700,527,.75),(1110,488,.65)]:
    ellipse((x-65*s,y-13*s,x+65*s,y+20*s),"papel")
    ellipse((x-35*s,y-33*s,x+15*s,y+12*s),"papel")
poly([(0,633),(170,521),(290,634),(440,549),(600,643),(900,610),(1160,580),(1600,670),(1600,942),(0,942)],"linha")
poly([(0,698),(200,613),(330,685),(500,609),(680,686),(940,650),(1260,677),(1600,625),(1600,942),(0,942)],"lima")
poly([(0,787),(270,703),(525,755),(820,709),(1050,760),(1400,693),(1600,740),(1600,942),(0,942)],"verde")
poly([(650,656),(694,656),(746,722),(676,764),(854,817),(955,942),(773,942),(715,854),(540,785),(615,724),(692,701)],"rio")
for x,y,s in [(70,697,.85),(131,703,1.05),(205,689,.8),(263,726,1.2),(345,710,.8),(388,719,1.15),
               (62,794,1.3),(151,831,1.35),(278,817,1.2),(391,853,1.5),(483,852,.9)]:
    tree(x,y,s,"floresta" if x%2 else "verde")
for x,y,s in [(829,681,.8),(947,673,.65),(1515,714,.9)]: turbine(x,y,s)
for x,y,w,h in [(1040,744,74,125),(1128,750,86,165),(1226,758,63,108)]: building(x,y,w,h)
house(1308,765,1.05)
house(1402,786,.8)
for x,y,s in [(971,811,.7),(1047,826,.8),(1134,842,.8)]: solar(x,y,s)
poly([(1117,942),(1103,918),(1240,865),(1420,854),(1600,829),(1600,857),(1440,884),(1260,891),(1160,942)],"menta")
for x,y,s in [(1380,871,.7),(1500,850,.6)]: person(x,y,s)
bike(1270,908,.6)
for x,y,s in [(1482,921,.8),(1524,924,.9),(1440,918,.65)]: tree(x,y,s,"floresta")
# Pequenas faixas de cultivo e marcas na água reforçam o detalhamento.
for j in range(5): line([(25+j*12,926),(170+j*12,882)],"lima",5)
for x,y in [(684,704),(624,771),(741,834),(821,903)]: line([(x-12,y),(x+19,y)],"ceu",3)
for x,y in [(434,529),(480,514),(1457,576)]:
    line([(x-9,y),(x,y+5),(x+9,y)],"tinta",2)
# Legendas posicionadas dentro de áreas tranquilas da ilustração.
txt(80,901,"BIODIVERSIDADE",18,"papel",True)
txt(790,907,"ÁGUA",18,"tinta",True)
txt(1120,500,"ENERGIA LIMPA",18,"tinta",True)
txt(1110,795,"CIDADES MAIS VERDES",17,"papel",True)
```

```python
# 5. Os três pilares: diagramas e sínteses, não percentuais.
print("3/4 • Compondo os pilares, os 5 Rs e as ações práticas...")
txt(80,993,"01",23,"verde",True)
txt(140,984,"O futuro depende de três pilares",44,bold=True)
line([(313,1122),(800,1122),(1287,1122)],"linha",4)
for x,kind,cor,fundo,titulo,desc in [
    (313,"folha","verde","menta","Ambiental","Proteger água, florestas e espécies.\nReduzir a poluição e o desperdício."),
    (800,"pessoas","azul","ceu","Social","Garantir saúde, educação e inclusão.\nPromover qualidade de vida para todos."),
    (1287,"economia","coral","rosa","Econômico","Gerar trabalho e renda com respeito\naos limites dos recursos naturais."),
]:
    circle(x,1122,74,fundo)
    icon(x,1122,kind,1.25,cor)
    txt(x,1215,titulo,35,cor,True,anchor="mt")
    # Quebra manual evita exceder a largura de cada coluna.
    linhas=desc.split("\n")
    for j,t in enumerate(linhas): txt(x,1270+j*35,t,22,"suave",anchor="mt")
```

```python
# 6. Fluxo dos 5 Rs: contraste, ícones grandes e orientações breves.
rect((0,1375,W,1775),"tinta")
txt(80,1417,"02",23,"ouro",True)
txt(140,1408,"Menos desperdício. Mais possibilidades.",43,"papel",True)
txt(80,1470,"OS 5 Rs DO CONSUMO CONSCIENTE",20,"lima",True)
rs=[("Repensar","repensar","Preciso mesmo?"),("Recusar","recusar","Evite o desnecessário."),
    ("Reduzir","reduzir","Consuma com cuidado."),("Reutilizar","reutilizar","Dê uma nova função."),
    ("Reciclar","reciclar","Separe corretamente.")]
for i,(nome,kind,nota) in enumerate(rs):
    x=208+i*296
    icon(x,1582,kind,1,"ouro")
    txt(x,1650,nome,30,"papel",True,anchor="mt")
    txt(x,1700,nota,20,"papel",anchor="mt")
    if i<4: seta((x+75,1582),(x+220,1582),"lima",3)
```

```python
# 7. Ações práticas: minicenários originais em quatro colunas.
txt(80,1823,"03",23,"verde",True)
txt(140,1814,"Pequenas escolhas, impacto coletivo",43,bold=True)
acoes=[("Água","agua","azul","Feche a torneira.\nConserte vazamentos."),
       ("Energia","energia","coral","Apague luzes ociosas.\nUse luz natural."),
       ("Mobilidade","bike","verde","Caminhe ou pedale.\nPrefira transporte coletivo."),
       ("Natureza","folha","verde","Cuide das áreas verdes.\nRespeite a biodiversidade.")]
for i,(titulo,kind,cor,desc) in enumerate(acoes):
    x=252+i*365
    circle(x,1960,67,"ceu" if i==0 else "rosa" if i==1 else "menta")
    icon(x,1960,kind,1.1,cor)
    txt(x,2052,titulo,34,cor,True,anchor="mt")
    for j,t in enumerate(desc.split("\n")): txt(x,2108+j*34,t,22,"suave",anchor="mt")
```
